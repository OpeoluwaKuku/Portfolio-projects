```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
ope = pd.read_excel(r'D:\Desktop\2025 ball stats.xlsx', sheet_name=['Ope'])
zubby = pd.read_excel(r'D:\Desktop\2025 ball stats.xlsx', sheet_name=['Zubby'])
```

```python
# ope is a dict returned by read_excel with sheet_name=['Ope']
# extract the actual DataFrame stored under the 'Ope' key and show its head
ope_df = ope['Ope']
ope_df.head()
```

```python
ope_df = ope_df.drop(columns= ['W','L','D', 'Unnamed: 9'])
ope_df.head()
```

```python
ope_df['Location'].value_counts()
```

### ADDING THE CATEGORY COLUMN
    The column is to differentiate the type of football pitch used.

```python
conditions = [ope_df['Location'].isin(['Peace fc', 'Thursday ball', 'Hommies fc']), ope_df['Location'].isin(['Gbaja', 'Razaq', 'Small london', 'Adisa Bashua', 'Kernel']), ope_df['Location'].isin(['Union Bank'])]
new_categories = ['5-aside mocking post coaltar', '5-aside full post', '11 aside full pitch']
ope_df['Category'] = np.select(conditions, new_categories)
ope_df['Category'] = ope_df['Category'].astype('category')
```

```python
#ope_df.drop(columns= ['W','L','D', 'Unnamed: 9'])
```

```python
col = ope_df.pop('Category')
ope_df.insert(2,'Category', col)
```

```python
ope_df.head()
```

### Creating the Cost Column


```python
ope_df['Cost'] = ope_df['Location'].map({'Kernel':2000,'Gbaja':3000, 'Razaq':2000, 'Hommies fc':600, 'Union Bank':4000,  'Thursday ball': 600, 'Adisa Bashua':0, 'Small london':1500,  'Peace fc':600})
```

```python
ope_df.groupby('Location')['Cost'].sum()
```

### Creating the Month column


```python
ope_df['Month'] = pd.Categorical(ope_df['Date'].dt.month_name(), 
                                 categories=
                                 ['January', 'February', 'March', 'April', 'May',  'June', 'July', 'August', 'September', 'October',  'November', 'December'], ordered=True)
```

```python
col2 = ope_df.pop('Month')
ope_df.insert(1,'Month', col2)
```
