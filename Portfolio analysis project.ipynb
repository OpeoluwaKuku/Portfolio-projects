```python
 import pandas as  pd


tickers = ['NVDA','GS', 'MSFT']
weights = pd.Series([0.4, 0.4, 0.2], index = tickers)

```

```python
weights
```

```python
prices = None

for t in tickers:
    h = yf.Tickers(t).history('max')[['Close']]#.rename(columns={'Close':t})
    prices = h if prices is None else prices.join(h, how='inner')
```

```python
prices
```

```python
asset_ret  = prices.pct_change().dropna()
```

```python
# Extract ticker columns from the MultiIndex and calculate weighted portfolio returns
asset_returns = asset_ret.xs("Close", axis=1, level="Price")
asset_returns = asset_returns.reindex(columns=weights.index)

returns = asset_returns.mul(weights, axis=1).sum(axis=1)


returns
```

```python
# Make both indexes timezone-naive before comparison
returns.index = returns.index.tz_localize(None)
bench_returns.index = bench_returns.index.tz_localize(None)

# Keep only dates available in both series
returns, bench_returns = returns.align(bench_returns, join="inner")

# Ensure both return series use identical, timezone-naive datetime indexes
returns.index = pd.to_datetime(returns.index).tz_localize(None)
bench_returns.index = pd.to_datetime(bench_returns.index).tz_localize(None)

returns, bench_returns = returns.align(bench_returns, join="inner")
returns = returns.sort_index()
bench_returns = bench_returns.sort_index()

# Pyfolio expects timezone-aware UTC indexes
returns.index = pd.to_datetime(returns.index, utc=True)
bench_returns.index = pd.to_datetime(bench_returns.index, utc=True)

# Keep only dates available in both series
returns, bench_returns = returns.align(
    bench_returns,
    join="inner",
)

returns = returns.sort_index()
bench_returns = bench_returns.sort_index()

pf.create_returns_tear_sheet(
    returns,
    live_start_date=pd.Timestamp("2023-01-01", tz="UTC"),
    benchmark_rets=bench_returns,
)
```
